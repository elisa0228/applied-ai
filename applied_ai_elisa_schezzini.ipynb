**Problem Domain: Scheduling**\
**1 Introduction**\
Scheduling is a central problem within the transportation domain, which includes flight scheduling, crew assignment, aircraft routing and maintenance planning while meeting operational, legal and safety requirements. Airline flight scheduling and airport slot allocation are among the most complex scheduling problems, as they involve large-scale combinatorial search spaces and multiple conflicting objectives [1].

Traditional optimisation techniques find it difficult to scale efficiently because of their NP-hard nature [2]. The global Airline Schedule Optimisation AI market was valued at USD 1.92 billion in 2024 and is projected to reach USD 7.18 billion by 2033, growing at a compound annual growth rate of 15.6% [3], reflecting the industries aggressive adoption of artificial intelligence to tackle these NP-hard problems. As a result, transportation scheduling has seen a rise in the use of artificial intelligence approaches, which provide scalable, adaptable and flexible solutions. This review explores how AI techniques have been applied to airline scheduling challenges and critically evaluates the strengths and limitations of these methods.

**2 Evolution of AI approaches in scheduling**\
AI-based scheduling methods have been widely adopted to address the shortcomings of optimisation techniques. Dynamic limitations, unpredictable demands and conflicting goals like robustness, efficiency and fairness are common. Early airline scheduling relied on rule-based systems and manual planning which requires months of planning per seasonal cycle.

A significant change occurred with the introduction of optimisation algorithms. Genetic algorithms became a prominent metaheuristic solution for crew scheduling problems, providing advantages over conventional methods in terms of computational efficiency [4]. However, due to poor inter-problem coordination, prior methods usually divided the scheduling problem into sequential sub-problems, such as aircraft routing, crew pairing and staff rostering, leading to global solutions [5].

Recent research has focused on integrated optimisation approaches. In comparison to traditional techniques Ouyang et al, [6] presented a parallel genetic algorithm framework for airline crew scheduling that achieved a 16.57-85.82% reduction in computation time while achieving a 97% flight-matched percentage and higher crew utilisation of nearly 44 flights per month.

AI-driven scheduling has been used in aviation for crew rostering, flight scheduling, airport slot allocation and disruption management. According to research, AI techniques routinely outperform conventional heuristics in crowded, large-scale transportation settings, especially where adaptation and flexibility are needed [7].

**3 AI techniques applied to airline scheduling**\
The airline scheduling area, has several AI and optimisation strategies that have been successfully applied which addresses different aspects of the scheduling challenge:

**3.1 Genetic Algorithms**\
The most popular metaheuristic method for transport scheduling issues, staff scheduling and aircraft routing. While Crego Hanne and Dornberger [8] showed multi-objective optimisation capabilities for crew management, Ouyang et al, [6] used parallel genetic algorithms to reduce computation time by 16.57-85.82%. Genetic Algorithms are excellent at managing several constraints at once and big combinatorial spaces.

In airline scheduling, GAs have been successfully to crew assignment, airport resource allocation and flight scheduling. They key advantage is offering flexibility in managing several goals, including operational efficiency, congestion reduction, and fairness [7]. Although GAs is effective, their stochastic nature needs robust evaluation across numerous runs, they also require careful parameter tweaking and may not guarantee optimal solutions [9].

**3.2 Machine Learning and Predictive Analytics**\
Machine learning techniques are increasingly used in the airline industry to support scheduling decisions through demand forecasting, delay prediction and schedule robustness analysis. These approaches do not directly solve scheduling optimisation problems, but instead provide predictive insights that inform decision-making.

**3.3 Real-time AI Decision Systems**\
Real-time AI decision systems are used to support operational scheduling adjustments during disruptions. For example, American Airlines flight hold system uses AI to make dynamic decisions about delaying departing flights to accommodate connecting passengers from delayed inbound aircraft [10]. These systems operate at the execution stage of scheduling rather than the planning stage, enhancing schedule robustness rather than generating schedules from scratch.

**3.4 Constraint Satisfaction Problem (CSP) Solvers**\
One of the most well-known AI scheduling methods is CSPs. Variables (such as aeroplane departure times), domains (such as available time slots), and to handle hard constraints in gate assignment, crew duty limitations and maintenance windows, conventional mathematical programming techniques are supplemented with AI enhanced heuristics [5], these are used to define the scheduling job in a CSP formulation. Strick constraint satisfaction is crucial for safety and compliance in airline scheduling and airport slot allocation, where CSPs have been effectively implemented [11].

The interpretability of CSPs and their capacity to ensure workable schedules that meet all given limitations are their main advantages. Constraint propagation and heuristic-guided backtracking are two methods that decrease the search space ad increase solver efficiency. However, as the number of variables and constrains rises, CSPs frequently experience scaling problems. Thousands of flights may be involved in large-scale airline scheduling issues, resulting in excessive computation times. To increase scalability and performance, CSPs are often coupled with heuristic or metaheuristic techniques [1].

**3.5 Heuristic and Greedy Scheduling Approaches**\
In transportation scheduling, heuristic and greedy scheduling techniques are frequently employed as starting points. These techniques are distributing resources according to priority rules, such as earliest departure time or first come first serves. Heuristics are frequently used in airline scheduling to provide generate initial workable schedules or to benchmark more sophisticated AI methods [12].

While heuristic methods are simple to use and computationally efficient, they don't usually consider global system performance, fairness concerns or long-term optimisation goals. As a result, they frequently produce low-quality timetables in congested transportation system. Rather than being a stand-alone solution, their main value is found in their use as comparison techniques or as parts of hybrid AI systems.

Each method has unique advantages and disadvantages, emphasising how crucial it is to choose the right AI strategies depending on the goals, constraints and objectives. **Technique 1: Genetic Algorithms (GA)**\
Natural evolution serves as the inspiration for Genetic Algorithms (GAs), a population-based metaheuristic that is frequently applied to challenging scheduling and optimisation issues. GAs have been used in airline scheduling to optimise goals like congestion reduction, delay minimisation and fairness across airlines in flight scheduling, crew scheduling and airport resource allocation [13, 14].

Since schedules may be recorded as chromosomes (integers for each flight) and assessed using a fitness function (rating of how good the schedule is), GAs are a good fit for the available airline scheduling data. The collection offers enough data to establish relevant fitness standards, including reducing departures that overlap or distributing slots evenly among airlines. In contrast to CSPs, GAs can function well even in situation where competing objectives need to be balanced, or restrictions are soft (preferences and goals but not mandatory).  

In airline scheduling, soft constraints are preferences that are desired but not required, such as preferring morning departures for business flights or aiming for turnaround time of less than 1.5 hours. Competing objectives happen when goals conflict: reducing fleet size (cutting costs) versus preserving backup capacity (operational resilience) or minimising aircraft turnaround time (for efficiency) versus minimising gate conflicts (for safety). For instance, a 2-hour buffer removes conflicts but requires additional aircraft, whereas a strict 45-minute turnaround maximises aircraft utilisation but increases the likelihood of gate conflicts. While CSPs treat constraints as binary (satisfied/violated) and cannot easily prioritise or compromise.

The setup time for a GA is higher than for heuristic methods, as it involves careful design of the chromosomal representation, fitness function and genetic operators (explore the search space and improve solutions via selection, crossover and mutation). However, once put into practice, GAs can efficiently grow to massive problem sizes and generate excellent answers in a reasonable amount of time. GAs are versatile in terms of computational cost because runtime may be regulated by population size and number of generations [9].

The output of a GA is an optimised schedule, rather than a workable timetable. Hence, GAs are especially suitable in situation where trade-offs and optimisation are crucial. However, GAs might produce inconsistent results across runs and do not guarantee optimal solution, necessitating repeated testing and assessment. Despite these limitations, GAs are ideal for real-world airline scheduling situations where flexibility and scalability are crucial.

**Technique 2: Constraint Satisfaction Problems (CSPs)**\
Constraint Satisfaction Problems are a well-known method for modelling scheduling issues where a strict follow of the constraints is required. CSPs have been widely used in airline scheduling for timetable feasibility checking and airport slot allocation, where variables represent flight departures, domains represent available time slots and constraints capture runway capacity, separation requirements and regulatory rules [11].

CSPs work well with structured scheduling data, like airline schedules, from the standpoint of data availability. Clearly specified departure times, routes and operation days are available in the Indian Domestic Airline Dataset. These can be discretised into time slots and modelled as CSP variables. Since CSPs mostly rely on constraint definitions rather than large datasets, they require relatively little historical data than machine learning techniques.

The time required to set up a CSP model is moderate to high, as constraints must be carefully formalised to reflect real-world operational rules. However, for small to medium issue sized, CSP solvers can provide correct schedules rather rapidly once they are defined. For large-scale airline scheduling problems, CSPs may experience long computation times as the search space grows exponentially [1].

When safety and regulatory compliance are non-negotiable, CSPs are especially ideal because their result is a workable timetable that meets all limitations. However, until they are expanded with optimisation layers, CSPs are less successful for optimisation goals like congestion minimisation or fairness. Therefore, CSPs work best when the main focus is on feasibility of when combined with optimisation strategies.

**Technique 3: Greedy and Heuristic Scheduling**\
Greedy and heuristic scheduling techniques allocate resources according to predefined rules, such as earliest departure time or first come first served. These methods are frequently used in airline scheduling as baselines or to create initial workable schedules that are compared to more sophisticated AI algorithms [12].

These approaches are quite compatible with the available dataset, due to their lack of historical learning data and minimum preprocessing requirements. Heuristic techniques are quick to apply and simple to comprehend because they require very little setup time. Additionally, they are computationally efficient because they generate findings nearly instantly, even for big datasets.

The primary limitation of heuristic scheduling is the quality of the output. Heuristics can provide workable schedules, but they typically don't take into consideration global optimisation goals like fairness or congestion reduction. Additionally, they find it difficult to adjust shifting demands or limitations, resulting in less-than-ideal solution in intricate and crowded settings.

Heuristic scheduling is therefore most suited for small scale issues, making decisions in real time, or serving as a benchmark for assessing more advanced AI methods. Heuristic scheduling offers a useful benchmark for evaluating the performance of optimisation based methods like GAs in this project.

Heuristic scheduling techniques are especially useful in real-time, time-sensitive operational scenarios when making decisions quickly is more crucial than finding globally optimal solutions. For instance, airlines frequently use greedy heuristics to swiftly reassign flights or crews based on priority rules during disruptions like weather delays or aircraft unavailability. These methods are suitable when solution speed is the only constraint, even if they may result in less-than-ideal global schedules due to their low computational cost. They are best utilised as initial solution generators or in conjunction with more sophisticated optimisation techniques like Genetic Algorithms because their reliance on local decision rules restricts their ability to account for system-wide interactions. This emphasises that heuristics approaches work best when response time or computational resources are limits, not when long-term schedule quality is the main goal.

**(c) - Prototype**

**Solution Representation**\
Each individual solution is represented as a chromosome, which is a list of integers.

Each gene corresponds to a flight and stores an assigned departure hour (0-23).

**Fitness Function**\
The fitness function evaluates how good a schedule is by applying penalties for:
*   Gate conflicts (same airport, same hour)
*   Undesirable nighttime departures (before 6am or after 10pm)

Conflicts are penalised to reflex operational importance.
The total penalty is negated so that higher fitness values represent better solutions, allowing the use of the max function during selection.

**Genetic Operators**
*   Selection: tournament selection (randomly choose three individual and return the best one)
*   Crossover: uniform crossover (each gene is chosen randomly from either parent)
*   Mutation: random reassignment of departure hour with low probability

These operators ensure exploration of the solution space (potential features) while keeping in mind the solution diversity.

**Prototype Execution**\
The algorithm runs for 50 generation with a population size of 20. The best fitness value from each generation is stores and visualised.

```python
#random sample of 200 flights from the clean dataset and random state of 42 so results become reproducible
ga_data = data.sample(n=200, random_state=42).reset_index(drop=True)

num_flights = len(ga_data) #number of flights is the chromosome length
```

```python
#create a candidate flight schedule, each gene represents the departure hour (0–23) for one flight
def create_individual(num_flights):
    return [random.randint(0, 23) for _ in range(num_flights)]
    #assign each flight a departure hour (0-23) and repeat for all flights
```

```python
#fitness function evaluates how good a schedule is, higher fitness = fewer conflicts and penalties
def fitness(schedule):
    conflicts = 0 #counter for gate conflicts
    airport_time_slots = {} #dictionary to track used airport hour slots

    #loop through each flight and its assigned departure hour
    for i, flight in enumerate(ga_data.itertuples()):
        origin = flight.origin #airport of departure
        dep_hour = schedule[i] #assigned departure hour for this flight

        key = (origin, dep_hour) #unique airport time identifier

        #check for gate conflicts (same airport, same hour)
        if key in airport_time_slots: #if slot is already used
            conflicts += 1 #count a conflict
        else:
            airport_time_slots[key] = True #otherwise reserve the slot

    #penalise undesirable night flights (soft constraint)
    night_penalty = 0
    for hour in schedule: #loop over all assigned hours
        if hour < 6 or hour > 22: #flights outside preferred operating hours
            night_penalty += 1 #add soft penalty (allowed but preferred to be daytime)

    #total penalty (conflicts are more important)
    total_penalty = (10 * conflicts) + night_penalty

    #return negative penalty because GA maximises fitness
    return -total_penalty
```

```python
#tournament selection: randomly choose 3 individuals and return the best one
def selection(population):
    return max(random.sample(population, 3), key=fitness)
```

```python
#uniform crossover: each gene is chosen randomly from either parent
def crossover(parent1, parent2):
    child = [] #new offspring chromosome (empty schedule)
    for i in range(len(parent1)): #go through each flight one by one
        if random.random() < 0.5: #50% chance from parent1
            child.append(parent1[i])
        else: #otherwise from parent2
            child.append(parent2[i])
    return child #return offspring (inherits traits from both parents)
```

```python
#randomly change departure hours with small probability
def mutate(individual, mutation_rate=0.05):
    for i in range(len(individual)):
        if random.random() < mutation_rate: #apply mutation with small probability (explores new solutions without destroying the good ones)
            individual[i] = random.randint(0, 23) #assign new random hour
    return individual
```

```python
population_size = 20 #create initial population
population = [create_individual(num_flights) for _ in range(population_size)]
```

```python
generations = 50
fitness_scores = [] #track best fitness per generation

for generation in range(generations):
    new_population = [] #new population container

    for _ in range(population_size):
        parent1 = selection(population) #select first parent
        parent2 = selection(population) #select second parent

        child = crossover(parent1, parent2) #recombine parents
        child = mutate(child) #mutate offspring

        new_population.append(child) #add child to new population

    population = new_population #replace old population

    #track best solution in this generation
    best_individual = max(population, key=fitness) #best solution in this generation
    fitness_scores.append(fitness(best_individual)) #store best solution
```

```python
plt.plot(fitness_scores) #plot fitness progression
plt.xlabel("Generation")
plt.ylabel("Fitness")
plt.title("Fitness per Generation (Genetic Algorithm Scheduling Optimisation)")
plt.show() #display the plot
```
